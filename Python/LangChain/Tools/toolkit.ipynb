```python
from langchain_core.tools import tool

# custom tools
@tool
def add(a:int,b:int)->int:
    """Add Two Number"""
    return a+b


@tool
def multiply(a:int,b:int)->int:
    """Multiply Two Number"""
    return a*b
```

```python
class mathTool:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit=mathTool()
tools=toolkit.get_tools()
```

```python
for tool in tools:
    print(tool.name,"=>",tool.description)
```
